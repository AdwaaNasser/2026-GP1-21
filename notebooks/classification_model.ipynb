# Burnout Level Classification

Same data cleaning and preprocessing as the regression model (`preprocessing.py`). Only the target (`burnout_level`) and the models are different.

```python
import sys
sys.path.append("..")             # preprocessing.py in the project root
```

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap
from scipy.stats import loguniform

from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             classification_report, ConfusionMatrixDisplay)

from preprocessing import DROP_COLS, SEED, build_preprocessor, load_data
```

## Load data

Same columns are dropped as in the regression model. `burnout_score` is also dropped here because `burnout_level` is made directly from it.

```python
df = load_data()

TARGET = "burnout_level"
X = df.drop(columns=DROP_COLS + ["burnout_score"])
y = df[TARGET]

print(X.shape)
y.value_counts(normalize=True).round(3)
```

## Train/test split

Same split as the regression model (80/20, stratified on `burnout_score` deciles, same seed), so both models are trained and tested on exactly the same employees.

```python
score_bins = pd.qcut(df["burnout_score"], q=10, labels=False, duplicates="drop")
X_train, X_test, y_train, y_test, score_train, score_test = train_test_split(
    X, y, df["burnout_score"], test_size=0.2, random_state=SEED, stratify=score_bins
)

print("train:", X_train.shape, " test:", X_test.shape)
pd.DataFrame({"train": y_train.value_counts(normalize=True),
              "test": y_test.value_counts(normalize=True)}).round(3)
```

```python
train_bins = pd.qcut(score_train, q=10, labels=False, duplicates="drop")
cv = list(StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED).split(X_train, train_bins))
```

## Preprocessing pipeline

Logic fixes → median imputation + scaling for numeric columns, most-frequent imputation + one-hot encoding for categorical columns.

```python
def make_model(estimator):
    return Pipeline([("prep", build_preprocessor(X_train)), ("model", estimator)])

prep = build_preprocessor(X_train).fit(X_train)
ct = prep.named_steps["columns"]
print("1. fix: logic rules (fix_logic)")
print("2. gender: missing or unknown gender -> Prefer not to say (clean_gender)")
for i, (name, pipe, cols) in enumerate(ct.transformers_[:2], start=3):
    steps = " -> ".join(type(step).__name__ for _, step in pipe.steps)
    print(f"{i}. {name} ({len(cols)} columns): {steps}")
```

## Model comparison (5-fold CV on train)

```python
models = {
    "Dummy (most frequent)": DummyClassifier(strategy="most_frequent"),
    "Decision Tree": DecisionTreeClassifier(random_state=SEED),
    "Naive Bayes": GaussianNB(),
    "Logistic Regression": LogisticRegression(max_iter=2000),
}
scoring = ["accuracy", "precision_macro", "recall_macro", "f1_macro"]

results = []
for name, est in models.items():
    s = cross_validate(make_model(est), X_train, y_train, cv=cv,
                       scoring=scoring, return_train_score=True)
    results.append({
        "model": name,
        "accuracy": s["test_accuracy"].mean(),
        "precision": s["test_precision_macro"].mean(),
        "recall": s["test_recall_macro"].mean(),
        "f1_macro": s["test_f1_macro"].mean(),
        "f1_std": s["test_f1_macro"].std(),
        "train_accuracy": s["train_accuracy"].mean(),
    })

results = pd.DataFrame(results).set_index("model").sort_values("f1_macro", ascending=False)
results.round(4)
```

Logistic Regression is clearly the best. The decision tree overfits (100% train accuracy), and Naive Bayes is held back by its independence assumption. We tune Logistic Regression.

## Hyperparameter tuning

```python
param_dist = {
    "model__C": loguniform(1e-3, 100),
    "model__class_weight": [None, "balanced"],
}
search = RandomizedSearchCV(make_model(LogisticRegression(max_iter=2000)), param_dist,
                            n_iter=20, cv=cv, scoring="f1_macro",
                            random_state=SEED, n_jobs=-1)
search.fit(X_train, y_train)

print("best params:", search.best_params_)
print("best CV macro F1:", round(search.best_score_, 4))
best_model = search.best_estimator_
```

## Final evaluation on the test set

```python
def evaluate(name, pred):
    return {"model": name,
            "accuracy": accuracy_score(y_test, pred),
            "precision": precision_score(y_test, pred, average="macro", zero_division=0),
            "recall": recall_score(y_test, pred, average="macro", zero_division=0),
            "f1_macro": f1_score(y_test, pred, average="macro")}

test_results = [
    evaluate(name, make_model(models[name]).fit(X_train, y_train).predict(X_test))
    for name in ["Dummy (most frequent)", "Decision Tree", "Naive Bayes"]
]
y_pred = best_model.predict(X_test)
test_results.append(evaluate("Logistic Regression (tuned)", y_pred))

pd.DataFrame(test_results).set_index("model").round(4)
```

```python
labels = ["Low", "Moderate", "High", "Severe"]
print(classification_report(y_test, y_pred, labels=labels, digits=3))
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, labels=labels, cmap="Blues", ax=ax)
ax.set_title("Logistic Regression - test set")
plt.show()
```

Most mistakes are between neighbouring levels (e.g. Moderate vs High), which makes sense since the levels are cut-offs on the same score.

## Regression vs classification

The regression model's scores can be turned into levels with the same cut-offs as `burnout_level` (≤ 3.5 Low, ≤ 5.5 Moderate, ≤ 7.0 High, above that Severe). Here we check whether this is as accurate as predicting the level directly, and the reverse: turning predicted levels into scores. The regression model is not retrained: cross-validation uses its tuned settings, and the test set uses the saved model.

```python
import joblib
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score


def score_to_level(scores):
    # same cut-offs as burnout_level in the dataset and in Bayyinah
    return np.select([scores <= 3.5, scores <= 5.5, scores <= 7.0], ["Low", "Moderate", "High"], "Severe")


def macro_f1(a, b):
    return f1_score(a, b, average="macro")


def fold_mean(pred, metric):
    # mean over the 5 folds, the same way cross_validate reports it
    return np.mean([metric(y_train.iloc[va], np.asarray(pred)[va]) for _, va in cv])


# tuned settings of the regression model (regression_model.ipynb)
tuned_hgb = HistGradientBoostingRegressor(
    learning_rate=0.0714737322617201, max_iter=951, max_leaf_nodes=23, max_depth=3,
    min_samples_leaf=116, l2_regularization=1.6313946190943742, max_features=0.8167648553804474,
    early_stopping=True, validation_fraction=0.1, n_iter_no_change=30, random_state=SEED)

reg_cv_levels = score_to_level(np.clip(cross_val_predict(make_model(tuned_hgb), X_train, score_train, cv=cv), 0, 10))
lr_cv_levels = cross_val_predict(best_model, X_train, y_train, cv=cv)

pd.DataFrame({"CV accuracy": [fold_mean(reg_cv_levels, accuracy_score), fold_mean(lr_cv_levels, accuracy_score)],
              "CV macro F1": [fold_mean(reg_cv_levels, macro_f1), fold_mean(lr_cv_levels, macro_f1)]},
             index=["Regression -> level", "Logistic Regression (tuned)"]).round(3)
```

```python
# test set: the saved regression model (the one used by Bayyinah) and the tuned Logistic Regression
reg_model = joblib.load("../burnout/ml/burnout_regression_model.joblib")
reg_test = np.clip(reg_model.predict(X_test), 0, 10)

# levels: regression scores cut into levels vs levels predicted directly
levels = pd.DataFrame({"accuracy": [accuracy_score(y_test, score_to_level(reg_test)), accuracy_score(y_test, y_pred)],
                       "macro F1": [macro_f1(y_test, score_to_level(reg_test)), macro_f1(y_test, y_pred)]},
                      index=["Regression -> level", "Logistic Regression (tuned)"])

# scores: each predicted level replaced by the mean training score of that level
level_means = score_train.groupby(y_train).mean()
lr_scores = pd.Series(y_pred).map(level_means).values
scores = pd.DataFrame({"MAE": [mean_absolute_error(score_test, reg_test), mean_absolute_error(score_test, lr_scores)],
                       "RMSE": [root_mean_squared_error(score_test, reg_test), root_mean_squared_error(score_test, lr_scores)],
                       "R2": [r2_score(score_test, reg_test), r2_score(score_test, lr_scores)]},
                      index=["HistGradientBoosting (tuned)", "Logistic Regression -> score"])
display(levels.round(3), scores.round(3))
```

Levels taken from the regression scores are as accurate as predicting the level directly (CV macro F1 0.693 vs 0.695; test accuracy 0.711 vs 0.710). The reverse does not hold: turning predicted levels into scores increases the MAE by about 32% (1.01 vs 0.76). One regression model therefore gives both the score and the level without losing accuracy, so regression is kept as the final approach.
